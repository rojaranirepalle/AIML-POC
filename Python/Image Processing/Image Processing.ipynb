## Image Processing

```python
# Import PIL and matplotlib
import matplotlib.pyplot as plt  # Visualization Library
from PIL import Image # Python Imageing Library
```

```python
# Open the image with help of module Image
flower = Image.open('flower.jpg')
```

```python
# Display the image
flower
```

```python
# Checck the data type
type(flower)
```

```python
# Converting to numpy arrays
import numpy as np
flower_array = np.asarray(flower)
flower_array
```

```python
# plotting the array in the axes
plt.imshow(flower_array)
```

```python
flower_array.shape
```
