```python
## Matplotlib
import matplotlib.pyplot as plt
import seaborn as sns #statistical visualizations

%matplotlib inline
plt.rcParams['figure.figsize'] = 8,4

import warnings
warnings.filterwarnings('ignore') #os error
```

```python
country_economy['InternetUsers']
```

```python
# Univariate analysis : plot the graph using 1 variable is called univariate analysis
# Bivariate analysis : plot the graph using 1 variable is called Bivariate analysis

visuals1 = sns.distplot(country_economy['InternetUsers'])  # Distribution plot
```

```python
visuals2 = sns.displot(country_economy['InternetUsers'])
```

```python
visuals2 = sns.displot(country_economy['InternetUsers'],bins=15)
```

```python
# Outliers : data points very far from the observations
# Outliers impacts the ML model
# So introduced probabilities.
# Outlier also called as anamoly detection.
visuals3 = sns.boxplot(data = country_economy, x= 'IncomeGroup', y ='BirthRate')
```

```python
visuals4 = sns.lmplot(data = country_economy, x= 'InternetUsers', y ='BirthRate')  # Linear model
```

```python
visuals5 = sns.lmplot(data = country_economy, x= 'InternetUsers', y ='BirthRate',fit_reg=False)
```

```python
visuals6 = sns.lmplot(data = country_economy, x= 'InternetUsers', y ='BirthRate',fit_reg=False,hue='IncomeGroup')
```

```python
visuals6 = sns.lmplot(data = country_economy, x= 'InternetUsers', y ='BirthRate',fit_reg=True,hue='IncomeGroup')
```
