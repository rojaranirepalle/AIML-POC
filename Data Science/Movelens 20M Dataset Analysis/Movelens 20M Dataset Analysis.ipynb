### Combine aggreagation, merging, and filters to get useful analytics

```python
average_rating = ratings_data.groupby('movieId',as_index=False).mean()
del average_rating['userId']
average_rating.head()
```

```python
box_office = movies_data.merge(average_rating, on = "movieId",how="inner")
box_office.tail()
```

```python
high_rated = box_office[box_office['rating'] >= 4.0]
high_rated.head()
```

```python
is_Adventure = box_office[box_office['genres'].str.contains('Adventure')]
is_Adventure.head()
```

```python
box_office[(box_office['genres'].str.contains('Adventure')) & (box_office['rating'] >= 4.0)].head()
```

```python
is_highly_rated = box_office['rating'] >= 4.0
is_Adventure = box_office['genres'].str.contains('Adventure')
box_office[is_Adventure & is_highly_rated][-5:]
```
